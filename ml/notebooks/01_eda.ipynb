```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('../data/dataset_cleaned.csv')
print(f"Dataset shape: {df.shape}")
df.head()
```

```python
# Class distribution
df['fault_type'].value_counts()
```

```python
# Grouped mean metrics per fault type
num_cols = [c for c in df.columns if c not in ['run_id', 'timestamp', 'fault_type']]
df.groupby('fault_type')[num_cols].mean().T
```
